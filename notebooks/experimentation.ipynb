# Preparação de Dados com Notas Sensoriais
Este notebook carrega a base raw completa e prepara um dataset que inclui variáveis sensoriais (`Aroma`, `Flavor`, `Aftertaste`, `Acidity`, `Body`, `Balance`, `Uniformity`, `Clean_Cup`, `Sweetness`).
O objetivo é gerar um arquivo processado com notas sensoriais que pode ser usado em modelos de classificação posteriores.

```python
import pandas as pd

# Carregamento da base original
df = pd.read_csv('../data/raw/arabica_coffee_full_table.csv')
```

```python
contagem_paises = df['Country_of_Origin'].value_counts()

paises_raros = contagem_paises[contagem_paises < 15].index.tolist()

df['Country_of_Origin'] = df['Country_of_Origin'].replace(
    paises_raros,
    'Outros'
)
```

```python
# Variáveis numéricas de produção
colunas_numericas = [
    'Moisture',
    'Category_One_Defects',
    'Category_Two_Defects',
    'Quakers',
    'Altitude'
]

# Variáveis categóricas
colunas_categoricas = [
    'Variety',
    'Processing_Method',
    'Country_of_Origin',
    'Color'
]

# Variáveis sensoriais
colunas_sensoriais = [
    'Aroma',
    'Flavor',
    'Aftertaste',
    'Acidity',
    'Body',
    'Balance',
    'Uniformity',
    'Clean_Cup',
    'Sweetness'
]

coluna_target = 'Total_Cup_Points'
```

```python
df_modelo = df[
    colunas_numericas +
    colunas_categoricas +
    colunas_sensoriais +
    [coluna_target]
].copy()
```

```python
df_modelo = df_modelo[df_modelo['Total_Cup_Points'] > 0]
```

```python
for coluna in colunas_numericas + colunas_sensoriais:
    df_modelo[coluna] = df_modelo[coluna].fillna(
        df_modelo[coluna].median()
    )
```

```python
for coluna in colunas_categoricas:
    df_modelo[coluna] = df_modelo[coluna].fillna('Unknown')
```

```python
def classificar(pontos):
    if pontos < 82.4:
        return "Tradicional"
    elif pontos < 84:
        return "Superior"
    else:
        return "Gourmet"
```

```python
df_modelo['Classe'] = df_modelo['Total_Cup_Points'].apply(classificar)
```

```python
print(df_modelo.isnull().sum())

print(df_modelo['Classe'].value_counts())

df_modelo.head()
```

```python
df_modelo.to_csv(
    '../data/processed/coffee_classificado_sensoriais.csv',
    index=False
)

print("Tabela salva com sucesso!")
```
